```python
def celcius_ke_fahrenheit(celcius):
    """Mengkonversi suhu dari Celcius ke Fahrenheit"""
    fahrenheit = (celcius * 9/5) + 32
    return fahrenheit

print(celcius_ke_fahrenheit(0))
print(celcius_ke_fahrenheit(100))
```

```python
def is_genap(bilangan_bulat):
    """Mengembalikan True jika bilangan genap, False jika ganjil"""
    if bilangan_bulat % 2 == 0:
        return True
    else:
        return False

print(is_genap(4))
print(is_genap(7))
```

```python
def nilai(skor):
    """Menampilkan keterangan 'LULUS' atau 'GAGAL' berdasarkan skor"""
    batas_lulus = 70
    if skor >= batas_lulus:
        print("LULUS")
    else:
        print("GAGAL")

nilai(80)
nilai(60)
```

```python
def bilangan(batas_atas):
    """Menampilkan bilangan ganjil kurang dari batas_atas"""
    ganjil = []
    for i in range(1, batas_atas):
        if i % 2 != 0:
            ganjil.append(i)

    print(ganjil, sep=",")

bilangan(20)
```
